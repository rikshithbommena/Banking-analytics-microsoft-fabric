```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

folder = r"C:\Users\riksh\OneDrive\Documents\data analyst Projects\Banking_project\banking_dataset_kaggle\data\csv"

customers = pd.read_csv(folder + r"\customers.csv")
accounts = pd.read_csv(folder + r"\accounts.csv")
loans = pd.read_csv(folder + r"\loans.csv")
cards = pd.read_csv(folder + r"\cards.csv")
transactions = pd.read_csv(folder + r"\transactions.csv")
merchants = pd.read_csv(folder + r"\merchants.csv")
branches = pd.read_csv(folder + r"\branches.csv")
```

```python
print("Customers:", customers.shape)
print("Accounts:", accounts.shape)
print("Loans:", loans.shape)
print("Cards:", cards.shape)
print("Transactions:", transactions.shape)
print("Merchants:", merchants.shape)
print("Branches:", branches.shape)
```

```python
Accounts['account_type'].value_counts().plot(kind='bar')

plt.title('Distribution of Account Types')
plt.xlabel('Account Type')
plt.ylabel('Number of Accounts')
plt.xticks(rotation=0)

plt.show()
```
